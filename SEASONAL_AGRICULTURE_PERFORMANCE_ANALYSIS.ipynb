```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")

from google.colab import files
from scipy.stats import f_oneway, kruskal, pearsonr
from sklearn.preprocessing import StandardScaler

uploaded = files.upload()
file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Dataset Shape:", df.shape)
display(df.head())

df = df.replace([np.inf, -np.inf], np.nan)
df = df.drop_duplicates().reset_index(drop=True)

print("\nMissing Values:")
display(df.isnull().sum().to_frame("Missing Values"))

target_columns = [
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Revenue_INR",
    "Profit_INR",
    "Water_Used_m3"
]

for col in target_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

for col in categorical_columns:
    if df[col].isnull().any():
        mode = df[col].mode()
        if len(mode) > 0:
            df[col] = df[col].fillna(mode.iloc[0])
        else:
            df[col] = df[col].fillna("Unknown")

print("\nDataset after cleaning:", df.shape)

if "Season" not in df.columns:
    raise ValueError("Season column is required for this project.")

print("\nSeason Counts:")
display(df["Season"].value_counts())

season_order = df["Season"].value_counts().index.tolist()

numeric_analysis_columns = [
    col for col in [
        "Yield_Tonnes_Ha",
        "Production_Tonnes",
        "Revenue_INR",
        "Profit_INR",
        "Water_Used_m3",
        "Water_Efficiency_t_per_1000m3",
        "Fertilizer_kg_ha",
        "Pesticide_Litre_ha",
        "Rainfall_mm",
        "Avg_Temperature_C",
        "Humidity_pct",
        "Soil_pH",
        "Soil_Moisture_pct",
        "Disease_Pest_Risk_pct"
    ]
    if col in df.columns
]

season_summary = df.groupby("Season")[numeric_analysis_columns].agg(
    ["mean", "median", "min", "max", "std"]
)

print("\n========== SEASONAL STATISTICAL SUMMARY ==========")
display(season_summary)

for col in numeric_analysis_columns:
    summary = (
        df.groupby("Season")[col]
        .mean()
        .sort_values(ascending=False)
    )

    print(f"\nAverage {col} by Season:")
    display(summary)

    plt.figure(figsize=(9, 5))
    sns.barplot(
        data=df,
        x="Season",
        y=col,
        order=season_order,
        estimator="mean"
    )
    plt.title(f"Average {col} Across Seasons")
    plt.xlabel("Season")
    plt.ylabel(col)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

if "Crop" in df.columns:

    crop_season = pd.crosstab(
        df["Season"],
        df["Crop"]
    )

    print("\n========== CROP DISTRIBUTION BY SEASON ==========")
    display(crop_season)

    plt.figure(figsize=(12, 6))
    crop_season.plot(
        kind="bar",
        figsize=(12, 6)
    )
    plt.title("Crop Distribution Across Seasons")
    plt.xlabel("Season")
    plt.ylabel("Number of Records")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

if (
    "Season" in df.columns
    and "Crop" in df.columns
    and "Yield_Tonnes_Ha" in df.columns
):

    crop_yield_season = (
        df.groupby(["Season", "Crop"])["Yield_Tonnes_Ha"]
        .mean()
        .reset_index()
    )

    print("\n========== YIELD BY SEASON AND CROP ==========")
    display(crop_yield_season)

    plt.figure(figsize=(12, 6))
    sns.barplot(
        data=crop_yield_season,
        x="Season",
        y="Yield_Tonnes_Ha",
        hue="Crop"
    )
    plt.title("Average Yield by Season and Crop")
    plt.xlabel("Season")
    plt.ylabel("Yield (Tonnes/Ha)")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

if "Irrigation_Method" in df.columns:

    irrigation_season = pd.crosstab(
        df["Season"],
        df["Irrigation_Method"]
    )

    print("\n========== IRRIGATION BY SEASON ==========")
    display(irrigation_season)

    plt.figure(figsize=(12, 6))
    irrigation_season.plot(
        kind="bar",
        figsize=(12, 6)
    )
    plt.title("Irrigation Methods Across Seasons")
    plt.xlabel("Season")
    plt.ylabel("Number of Farms")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

if (
    "Season" in df.columns
    and "Irrigation_Method" in df.columns
    and "Yield_Tonnes_Ha" in df.columns
):

    irrigation_yield = (
        df.groupby(
            ["Season", "Irrigation_Method"]
        )["Yield_Tonnes_Ha"]
        .mean()
        .reset_index()
    )

    print("\n========== IRRIGATION VS YIELD ==========")
    display(irrigation_yield)

    plt.figure(figsize=(12, 6))
    sns.barplot(
        data=irrigation_yield,
        x="Season",
        y="Yield_Tonnes_Ha",
        hue="Irrigation_Method"
    )
    plt.title("Yield by Irrigation Method and Season")
    plt.xlabel("Season")
    plt.ylabel("Yield (Tonnes/Ha)")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

if "Rainfall_mm" in df.columns and "Yield_Tonnes_Ha" in df.columns:

    plt.figure(figsize=(9, 6))

    sns.scatterplot(
        data=df,
        x="Rainfall_mm",
        y="Yield_Tonnes_Ha",
        hue="Season"
    )

    plt.title("Rainfall vs Yield Across Seasons")
    plt.xlabel("Rainfall (mm)")
    plt.ylabel("Yield (Tonnes/Ha)")
    plt.tight_layout()
    plt.show()

if "Fertilizer_kg_ha" in df.columns and "Yield_Tonnes_Ha" in df.columns:

    plt.figure(figsize=(9, 6))

    sns.scatterplot(
        data=df,
        x="Fertilizer_kg_ha",
        y="Yield_Tonnes_Ha",
        hue="Season"
    )

    plt.title("Fertilizer Usage vs Yield Across Seasons")
    plt.xlabel("Fertilizer (kg/ha)")
    plt.ylabel("Yield (Tonnes/Ha)")
    plt.tight_layout()
    plt.show()

if "Water_Used_m3" in df.columns and "Yield_Tonnes_Ha" in df.columns:

    plt.figure(figsize=(9, 6))

    sns.scatterplot(
        data=df,
        x="Water_Used_m3",
        y="Yield_Tonnes_Ha",
        hue="Season"
    )

    plt.title("Water Usage vs Yield Across Seasons")
    plt.xlabel("Water Used (m³)")
    plt.ylabel("Yield (Tonnes/Ha)")
    plt.tight_layout()
    plt.show()

if "Season" in df.columns and "Profit_INR" in df.columns:

    season_profit = (
        df.groupby("Season")["Profit_INR"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\n========== PROFIT BY SEASON ==========")
    display(season_profit)

    best_profit_season = season_profit.idxmax()
    worst_profit_season = season_profit.idxmin()

    print(
        "Highest average profit season:",
        best_profit_season
    )

    print(
        "Lowest average profit season:",
        worst_profit_season
    )

if "Season" in df.columns and "Revenue_INR" in df.columns:

    season_revenue = (
        df.groupby("Season")["Revenue_INR"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\n========== REVENUE BY SEASON ==========")
    display(season_revenue)

if "Season" in df.columns and "Production_Tonnes" in df.columns:

    season_production = (
        df.groupby("Season")["Production_Tonnes"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\n========== PRODUCTION BY SEASON ==========")
    display(season_production)

if (
    "Season" in df.columns
    and "Disease_Pest_Risk_pct" in df.columns
):

    risk_season = (
        df.groupby("Season")["Disease_Pest_Risk_pct"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\n========== DISEASE/PEST RISK BY SEASON ==========")
    display(risk_season)

    plt.figure(figsize=(9, 5))
    sns.barplot(
        x=risk_season.index,
        y=risk_season.values
    )
    plt.title("Average Disease/Pest Risk by Season")
    plt.xlabel("Season")
    plt.ylabel("Risk (%)")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

if "Season" in df.columns and "Water_Used_m3" in df.columns:

    water_season = (
        df.groupby("Season")["Water_Used_m3"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\n========== WATER USAGE BY SEASON ==========")
    display(water_season)

if (
    "Season" in df.columns
    and "Water_Efficiency_t_per_1000m3" in df.columns
):

    efficiency_season = (
        df.groupby("Season")
        ["Water_Efficiency_t_per_1000m3"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\n========== WATER EFFICIENCY BY SEASON ==========")
    display(efficiency_season)

if "Season" in df.columns and "Rainfall_mm" in df.columns:

    rainfall_season = (
        df.groupby("Season")["Rainfall_mm"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\n========== RAINFALL BY SEASON ==========")
    display(rainfall_season)

if "Season" in df.columns and "Avg_Temperature_C" in df.columns:

    temperature_season = (
        df.groupby("Season")["Avg_Temperature_C"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\n========== TEMPERATURE BY SEASON ==========")
    display(temperature_season)

if "Season" in df.columns and "Humidity_pct" in df.columns:

    humidity_season = (
        df.groupby("Season")["Humidity_pct"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\n========== HUMIDITY BY SEASON ==========")
    display(humidity_season)

if "Season" in df.columns and "Soil_pH" in df.columns:

    soil_ph_season = (
        df.groupby("Season")["Soil_pH"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\n========== SOIL pH BY SEASON ==========")
    display(soil_ph_season)

if (
    "Season" in df.columns
    and "Soil_Moisture_pct" in df.columns
):

    soil_moisture_season = (
        df.groupby("Season")["Soil_Moisture_pct"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\n========== SOIL MOISTURE BY SEASON ==========")
    display(soil_moisture_season)

if "Season" in df.columns and "Fertilizer_kg_ha" in df.columns:

    fertilizer_season = (
        df.groupby("Season")["Fertilizer_kg_ha"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\n========== FERTILIZER USAGE BY SEASON ==========")
    display(fertilizer_season)

if "Season" in df.columns and "Pesticide_Litre_ha" in df.columns:

    pesticide_season = (
        df.groupby("Season")["Pesticide_Litre_ha"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\n========== PESTICIDE USAGE BY SEASON ==========")
    display(pesticide_season)

if "Season" in df.columns and "Yield_Tonnes_Ha" in df.columns:

    groups = [
        group["Yield_Tonnes_Ha"].dropna().values
        for _, group in df.groupby("Season")
    ]

    groups = [
        g for g in groups
        if len(g) > 1
    ]

    if len(groups) >= 2:

        anova_stat, anova_p = f_oneway(*groups)

        print("\n========== ANOVA TEST ==========")
        print("F-statistic:", round(anova_stat, 4))
        print("P-value:", round(anova_p, 6))

        if anova_p < 0.05:
            print(
                "Result: Yield differs significantly across seasons."
            )
        else:
            print(
                "Result: No statistically significant yield difference was detected across seasons."
            )

if "Season" in df.columns:

    print("\n========== KRUSKAL-WALLIS TEST ==========")

    for col in numeric_analysis_columns:

        groups = [
            group[col].dropna().values
            for _, group in df.groupby("Season")
        ]

        groups = [
            g for g in groups
            if len(g) > 1
        ]

        if len(groups) >= 2:

            try:

                statistic, p_value = kruskal(*groups)

                print(
                    f"{col}: statistic={statistic:.4f}, "
                    f"p-value={p_value:.6f}"
                )

            except Exception:
                pass

correlation_columns = [
    col for col in [
        "Rainfall_mm",
        "Avg_Temperature_C",
        "Humidity_pct",
        "Soil_pH",
        "Soil_Moisture_pct",
        "Nitrogen_kg_ha",
        "Phosphorus_kg_ha",
        "Potassium_kg_ha",
        "Fertilizer_kg_ha",
        "Pesticide_Litre_ha",
        "Water_Used_m3",
        "Disease_Pest_Risk_pct"
    ]
    if col in df.columns
]

if "Yield_Tonnes_Ha" in df.columns:

    correlation_results = []

    for col in correlation_columns:

        temp = df[
            [col, "Yield_Tonnes_Ha"]
        ].dropna()

        if len(temp) > 2:

            try:

                corr, p_value = pearsonr(
                    temp[col],
                    temp["Yield_Tonnes_Ha"]
                )

                correlation_results.append({
                    "Feature": col,
                    "Correlation": corr,
                    "P_Value": p_value
                })

            except Exception:
                pass

    correlation_df = pd.DataFrame(
        correlation_results
    ).sort_values(
        by="Correlation",
        ascending=False
    )

    print("\n========== CORRELATION WITH YIELD ==========")
    display(correlation_df)

if len(numeric_columns) > 1:

    plt.figure(figsize=(16, 10))

    sns.heatmap(
        df[numeric_columns].corr(),
        cmap="coolwarm",
        center=0
    )

    plt.title("Correlation Heatmap of Agricultural Variables")
    plt.tight_layout()
    plt.show()

print("\n============================================")
print("AUTOMATIC FINDINGS")
print("============================================")

if "Yield_Tonnes_Ha" in df.columns:

    yield_season = (
        df.groupby("Season")["Yield_Tonnes_Ha"]
        .mean()
        .sort_values(ascending=False)
    )

    best_yield = yield_season.idxmax()
    worst_yield = yield_season.idxmin()

    print(
        f"1. Highest average yield was observed in "
        f"{best_yield}."
    )

    print(
        f"2. Lowest average yield was observed in "
        f"{worst_yield}."
    )

if "Profit_INR" in df.columns:

    profit_season = (
        df.groupby("Season")["Profit_INR"]
        .mean()
        .sort_values(ascending=False)
    )

    print(
        f"3. Highest average profit was observed in "
        f"{profit_season.idxmax()}."
    )

    print(
        f"4. Lowest average profit was observed in "
        f"{profit_season.idxmin()}."
    )

if "Rainfall_mm" in df.columns:

    rainfall = (
        df.groupby("Season")["Rainfall_mm"]
        .mean()
        .sort_values(ascending=False)
    )

    print(
        f"5. Highest average rainfall was observed in "
        f"{rainfall.idxmax()}."
    )

if "Water_Used_m3" in df.columns:

    water = (
        df.groupby("Season")["Water_Used_m3"]
        .mean()
        .sort_values(ascending=False)
    )

    print(
        f"6. Highest average water usage was observed in "
        f"{water.idxmax()}."
    )

if "Disease_Pest_Risk_pct" in df.columns:

    risk = (
        df.groupby("Season")["Disease_Pest_Risk_pct"]
        .mean()
        .sort_values(ascending=False)
    )

    print(
        f"7. Highest average disease/pest risk was observed in "
        f"{risk.idxmax()}."
    )

if "Fertilizer_kg_ha" in df.columns:

    fertilizer = (
        df.groupby("Season")["Fertilizer_kg_ha"]
        .mean()
        .sort_values(ascending=False)
    )

    print(
        f"8. Highest average fertilizer usage was observed in "
        f"{fertilizer.idxmax()}."
    )

if "Water_Efficiency_t_per_1000m3" in df.columns:

    efficiency = (
        df.groupby("Season")
        ["Water_Efficiency_t_per_1000m3"]
        .mean()
        .sort_values(ascending=False)
    )

    print(
        f"9. Highest water efficiency was observed in "
        f"{efficiency.idxmax()}."
    )

print("\n============================================")
print("CONCLUSION")
print("============================================")

print(
    "The seasonal analysis identifies meaningful differences "
    "in agricultural performance across seasons."
)

print(
    "Yield, production, economic outcomes, environmental "
    "conditions and resource usage were compared across seasons."
)

print(
    "The statistical analysis helps determine whether observed "
    "seasonal differences are statistically meaningful."
)

print(
    "The visualizations provide evidence for identifying "
    "seasonal trends, relationships and unusual patterns."
)

print(
    "These findings can support evidence-based seasonal "
    "agricultural planning."
)

print("\n============================================")
print("DATA-DRIVEN RECOMMENDATIONS")
print("============================================")

print(
    "1. Select suitable crops based on observed seasonal yield patterns."
)

print(
    "2. Optimize irrigation according to seasonal water requirements."
)

print(
    "3. Monitor rainfall, temperature and humidity when planning cultivation."
)

print(
    "4. Use fertilizer according to crop and seasonal requirements."
)

print(
    "5. Monitor disease and pest risk during seasons with higher observed risk."
)

print(
    "6. Compare seasonal profitability before making agricultural decisions."
)

print(
    "7. Improve water efficiency by comparing water usage with yield."
)

print(
    "8. Use regional and seasonal patterns to support future agricultural planning."
)

cleaned_file = "cleaned_seasonal_agriculture_dataset.csv"

df.to_csv(
    cleaned_file,
    index=False
)

summary_file = "seasonal_agriculture_summary.csv"

season_summary.to_csv(
    summary_file
)

print("\nFiles created:")
print(cleaned_file)
print(summary_file)

files.download(cleaned_file)
files.download(summary_file)

print("\nPROJECT ANALYSIS COMPLETED SUCCESSFULLY.")
```
